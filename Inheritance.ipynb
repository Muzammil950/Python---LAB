```python
class test:

    def test_methd(self):
        return "this is first inheritence"
```

```python
class child_test(test):
    pass
```

```python
class_test_obj = child_test()
```

```python
class_test_obj.test_methd()
```

```python
# Multilevel Inheritance
```

```python
class class1:

    def test_class1(self):
        return "this is from class1"
```

```python
class class2(class1):

    def test_class2(self):
        return "this is from class2"
```

```python
class class3(class2):
    pass
        
```

```python
class_obj3 = class3()
```

```python
class_obj3.test_class1()
```

```python
class_obj3.test_class2()
```

```python
# Multiple Inheritance
```

```python
class class1:

    def test_class1(self):
        return "this is class 1"
```

```python
class class2:

    def test_class2(self):
        return "this is class 2"
```

```python
class class3(class1, class2):
    pass
```

```python
obj_class3 = class3()
```

```python
obj_class3.test_class1()
```

```python
obj_class3.test_class2()
```
