```python
# creating class
class student:

    #constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # method
    def display(self):
        print('name:',self.name)
        print('name:',self.age)
```

```python
# accessing/calling method
rohan.display()
```

```python
# class
class car :

    #constructor
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    # method
    def start(self):
        return self.brand, self.model
    
    
```

```python
#creating object
car1 = car("toyota","innova")
```

```python
# accessing method
car1.start()
```

```python
# class
class student:

    #constructor
    def __init__(self,name, age, marks):
        self.name = name
        self.age = age 
        self.marks = marks

    # methed
    def display(self):
        return self.name, self.age, self.marks

    # method 
    def result(self):
        if self.marks >= 40:
            return "pass"
        else :
            return "fail"
```

```python
# creating object
sohan = student('muzz',22,60)
```

```python
sohan.display()
```

```python
sohan.result()
```
