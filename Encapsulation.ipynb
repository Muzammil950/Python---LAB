```python
class car:

    def __init__(self, year, make, model, speed):
        self.__year = year
        self.__make = make
        self.__model = model
        self.__speed = 12

    def set_speed(self, speed):
        self.__speed = 0 if speed < 0 else speed

    def get_speed(self):
        return self.__speed
```

```python
c = car(2021, "toyota", "innova", 12)
```

```python
c._car__year
```

```python
c.set_speed(-2322)
```

```python
c.get_speed()
```

```python
c.set_speed(2345)
```

```python
c.get_speed()
```

```python
 class bank_account:

     def __init__(self, balance):
         self.__balance = balance

     def deposite(self, amount):
         self.__balance = self.__balance + amount

     def withdraw(self, amount):
         if self.__balance >= amount:
              self.__balance = self.__balance - amount
              return True
         else:
              return False

     def get_balance(self):
          return self.__balance
```

```python
muzz = bank_account(1000)
```

```python
muzz.get_balance()
```

```python
muzz.deposite(5000)
```

```python
muzz.get_balance()
```

```python
muzz.withdraw(7000)
```

```python
muzz.withdraw(3000)
```

```python
muzz.get_balance()
```
