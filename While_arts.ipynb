# WHILE LOOP

```python
a = 1
while a <= 10:
    print(a)
    a = a + 1
```

```python
n = int(input("enter your limit"))
starting_point = 0
counter = 1

while counter <= n:
    starting_point = starting_point + counter
    counter = counter + 1

starting_point
```

```python
number = int(input("enter your number : "))

factorial = 1
while number > 0:
    factorial = factorial * number
    number = number - 1
factorial
```

```python
number = int(input("enter your number :"))
a, b = 0, 1
counter = 0
while counter < number:
    print(a)
    c = a + b
    a = b
    b = c
    counter = counter + 1
```

```python
a, b = 0, 1

for i in range(5):
    print(a)
    c = a + b
    a = b
    b = c
```

```python
word = input("enter : ")
```

```python
reverse = " "
length = len(word)
while length > 0 :
    reverse = reverse + word[length - 1]
    length = length -1 
print(reverse)
```

```python
n = int(input("enter your number : "))
i = 1
while i <= 10 :
    result = n * i
    print(n, "* ", i ,"=" ,result)
    i = i + 1
          
```

```python
n = 5
i = 1
while i < n :
    print(i)
    i = i + 1
else :
    print("loop is completed")
```

```python
n = 5
i = 1
while i < n:
    print(i)
    if i == 3 :
        break
    i = i + 1
else:
    print("loop is complete")
```
